### 1. A class consists of n=24 students. The average BMI for the class is 32.46, and the standard deviation is 6.92. Six new students have joined the class. Their BMIs are [23.5, 30.5, 42.3, 34.5, 49.5, 44.2]. Write Python code and calculate the sample mean  and standard deviation for the 30 students using the recursive formula.

```python
from math import sqrt


new_bmis = [23.5, 30.5, 42.3, 34.5, 49.5, 44.2]
sample_mean, sample_std, n_students = 32.46, 6.92, 24
Q = sample_std ** 2 * n_students
while new_bmis:
    n_students += 1
    bmi_i = new_bmis.pop(0)
    last_sample_mean = sample_mean
    sample_mean = last_sample_mean + ((bmi_i-last_sample_mean)/float(n_students))
    Q = Q + (bmi_i - (last_sample_mean))*(bmi_i-sample_mean)

! finish the standard deviation equation outside of the do loop
sample_std = sqrt(Q/float(n_students))
print("sample mean:", sample_mean)
print("sample standard deviation:", sample_std)
```

### You shuffle a standard deck of playing cards, and deal a hand of 10 cards. With what probability does this hand have five red cards?

_First of all, this question NEEDS clarification. Is the question asking the chances of getting exactly five red cards or at least five red cards? The formula for finding the chances of exactly five red cards is as follows:_

The total number of ways to draw 10 cards out of 52 is C(52, 10). The number of ways to draw 5 red cards out of 26 is C(26, 5). The number of ways to draw the remaining 5 cards out of the 26 non-red cards is C(26, 5).

The probability is therefore:

P = [C(26, 5) * C(26, 5)] / C(52, 10)

If we just need to draw AT LEAST five cards, we would have to do extra calculations, shown below

```python
from math import comb


exactly_5_cards = comb(26, 5) * comb(26, 5) / comb(52, 10)
print(f"exactly 5 red cards: {exactly_5_cards*100}%")
```

```python
at_least_5_cards = 0
for num_red_needed in range(5, 11):
    at_least_5_cards += comb(26, num_red_needed) * comb(26, 10 - num_red_needed) / comb(52, 10)
print(f"at least 5 red cards: {at_least_5_cards*100}%")
```
