```python
data = [
    [0.01, 0.01, 0.03],
    [0.03, 0.08, 0.07],
    [0.03, 0.06, 0.06],
    [0.07, 0.07, 0.13],
    [0.12, 0.04, 0.03],
    [0.08, 0.07, 0.01],
]
```

```python
R = {}
for x in range(6):
    for y in range(3):
        r = 3 * x + 9 * y
        R[r] = R.get(r, 0.0) + data[x][y]

R = {k: round(v, 2) for k, v in sorted(R.items())}
R
```

```python
import matplotlib.pyplot as plt
plt.plot(R.keys(), R.values())
plt.show()
```

```python
expected_revenue = sum([revenue*probability for revenue, probability in R.items()])
print("Expected Revenue:", expected_revenue)
print("Expected Profit:", round(expected_revenue-10, 2))
```

```python
from math import sqrt

variance_revenue = sum((r ** 2) * p for r, p in R.items())- expected_revenue**2
sqrt(variance_revenue)
```
