```python
import numpy as np
import matplotlib.pyplot as plt

def cumulative_distribution_function(t: float) -> float:
    if t < 1:
        return 0
    elif 1 <= t < 3:
        return 1/8
    elif 3 <= t < 5:
        return 3/8
    elif 5 <= t < 7:
        return 7/8
    else:
        return 1

t_values = np.linspace(0, 7, 1000)
F_values = [cumulative_distribution_function(t) for t in t_values]
plt.figure(figsize=(8, 5))
plt.step(t_values, F_values, where='post')
plt.xlabel('t')
plt.ylabel('F(t)')
plt.title('CDF of T')
plt.xticks([0, 1, 3, 5, 7, 8])
plt.yticks([0, 1/8, 3/8, 7/8, 1])
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

# 2. The data “student-por.csv” consists of Portuguese language course students final grades (column “G3”) in two schools “GP” (Gabriel Pereira) and “MS” (Mousinho da Silveira). Let X and Y stand for the grade “G3” in the two schools respectively. Note that X and Y are discrete variables taking possible values 0, 1, 2, . . . , 20. 

```python
import pandas as pd
from pandas import read_csv
import numpy as np
import matplotlib.pyplot as plt


def compute_distribution(df, school):
    grades = df[df['school'] == school]['G3']
    distribution = grades.value_counts(normalize=True).sort_index()
    return distribution

def plot_distributions(dist_x, dist_y, title):
    plt.figure(figsize=(10, 5))
    if dist_x is not None:
        plt.bar(dist_x.index, dist_x.values, alpha=0.6, label='GP', color='blue')
    if dist_y is not None:
        plt.bar(dist_y.index, dist_y.values, alpha=0.6, label='MS', color='red')
    plt.xlabel('Final Grade')
    plt.ylabel('Probability')
    plt.title(title)
    plt.legend()
    plt.show()

def compute_statistics(dist):
    values = np.array(dist.index)
    probabilities = np.array(dist.values)
    mean = np.sum(values * probabilities)
    variance = np.sum((values - mean) ** 2 * probabilities)
    std_dev = np.sqrt(variance)
    return mean, std_dev

def main():
    df = read_csv('student-por.csv' )
    
    dist_x = compute_distribution(df, 'GP')
    dist_y = compute_distribution(df, 'MS')
    
    print("(a)")
    plot_distributions(None, dist_y, 'Grade Distributions of GP')
    plot_distributions(dist_x, None, 'Grade Distributions of MS')
    plot_distributions(dist_x, dist_y, 'Grade Distributions of GP and MS')
    
    mean_x, std_x = compute_statistics(dist_x)
    mean_y, std_y = compute_statistics(dist_y)
    
    print("(b)")
    print(f'GP: Mean = {mean_x:.2f}, Std Dev = {std_x:.2f}')
    print(f'MS: Mean = {mean_y:.2f}, Std Dev = {std_y:.2f}')
    
    print("(c)")
    if mean_x > mean_y:
        print("GP students performed better on average.")
    elif mean_x < mean_y:
        print("MS students performed better on average.")
    else:
        print("Both schools performed similarly on average.")

main()
```

- **Total Loss (100%)**: 
   - Loss = $400,000
   - Probability = 0.002
   - Expected Loss = \( 0.002 \times 400,000 = 800 \)

- **50% Loss**:
   - Loss = 50% of $400,000 = $200,000
   - Probability = 0.01
   - Expected Loss = \( 0.01 \times 200,000 = 2,000 \)

- **25% Loss**:
   - Loss = 25% of $400,000 = $100,000
   - Probability = 0.1
   - Expected Loss = \( 0.1 \times 100,000 = 10,000 \)

Now, sum up the expected losses:

Total Expected Loss = 800 + 2,000 + 10,000 = 12,800



Let P be the premium charged by the insurance company. The company wants an average profit of $500, so the expected premium income must cover the expected loss plus the desired profit.

Thus, the equation for the premium is:

P = Expected Loss + Desired Profit

Substitute the known values:

P = 12,800 + 500 = 13,300

The insurance company should charge a premium of **$13,300** per year to realize an average profit of $500.
