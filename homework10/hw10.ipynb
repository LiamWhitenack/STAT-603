(a) Let X be the number of calls that arrive at the call center in a 10 minute interval.
Find the observed probability (relative frequency) distribution of X from the data
collected in October.

```python
from collections import Counter


observed_distribution = {
    k: v / len(call_center_data) for k, v in sorted(Counter(call_center_data).items())
}
observed_distribution
```

(c) Randomly occurring telephone calls are often modeled by a Poisson distribution. Calculate the theoretical probability distribution of X assuming a Poisson distribution with mean λ that you calculated in part (b).

```python
from scipy.stats import poisson


poisson_distribution = {
    k: poisson.pmf(k, mu=mean(call_center_data)) for k in observed_distribution
}
poisson_distribution
```
