##### (a) Consider the Birthday Problem. Modify the Jupyter notebook to calculate using simulations the probability in a group of n people exactly two people share a birthday. Calculate the probability using 1 million simulations for n = 10.

```python
from random import randint

n = 10
only_two_share_birthdays_count = 0 
for _ in range(1_000_000):
    if len(set([randint(1, 365) for __ in range(n)])) == 9:
        only_two_share_birthdays_count += 1

print(only_two_share_birthdays_count / 1_000_000)
```

##### (b) Derive a formula for the true probability in a group of n people exactly two people share a birthday. Calculate the true probability for n = 10. Compare your answer with the one obtained using simulations in (a).

```python
import math

def probability_exactly_two_share_birthday(n) -> float:
    if n < 2:
        return 0

    choose_two: int = (n * (n - 1)) // 2
    shared_birthday_prob: float = 1 / 365
    unique_birthdays_prob: float = math.prod([(364 - i) / 365 for i in range(n - 2)])
    return choose_two * shared_birthday_prob * unique_birthdays_prob

print(probability_exactly_two_share_birthday(10))
```

The answer is nearly the same as it was when using a monte carlo simulation.

##### (c) Calculate the minimum number of people in a room for the above probability to exceed 20%.

```python
i = 0
probability = probability_exactly_two_share_birthday(i)
while probability <= 0.2:
    i += 1
    probability = probability_exactly_two_share_birthday(i)

print(i, probability)
```

The answer is 14.
