```python
import numpy as np
import scipy.stats as statistics
import matplotlib.pyplot as plot

total_reservations = 43
plane_seats = 40

probability_values = np.array([i / 100 for i in range(70, 91)])

probability_three_empty = []
probability_one_or_more_empty = []

for probability in probability_values:
    binomial_distribution = statistics.binom(total_reservations, probability)

    probability_37 = binomial_distribution.pmf(37)
    probability_three_empty.append(probability_37)

    probability_at_most_40 = binomial_distribution.cdf(40)
    probability_one_or_more_empty.append(probability_at_most_40)

best_probability_a = probability_values[np.argmax(probability_three_empty)]

probability_b_less_95 = probability_values[
    np.array(probability_one_or_more_empty) < 0.95
]

# Print results
print(
    f"(a): \n"
    + f"Probability of missing 3 seats for each p: {[round(float(value), 3) for value in probability_three_empty]}\n"
    + f"Maximum probability of exactly 3 empty seats occurs at probability = {best_probability_a:.3f}\n"
)
print(
    f"(b): \n"
    + f"Probabilities of having at least one empty seat: {[f'{probability_values[i]}: {round(float(value), 3)}' for i, value in enumerate(probability_one_or_more_empty)]}\n"
    + f"Probability values where probability of at least one empty seat < 0.95: {probability_b_less_95}\n"
)
```

```python
# Hypergeometric

# a. P(X = 4) = (C(7,4) * C(3,0)) / C(10,4)  
# b. P(X ≥ 2) = 1 - P(X = 0) - P(X = 1)

from math import comb

def hypergeom_prob(success, total_success, sample, total):
    return (comb(total_success, success) * comb(total - total_success, sample - success)) / comb(total, sample)

p_all_fire = hypergeom_prob(4, 7, 4, 10)
p_at_most_2_fail = 1 - hypergeom_prob(0, 7, 4, 10) - hypergeom_prob(1, 7, 4, 10)

p_all_fire, p_at_most_2_fail
```

```python
# Poisson

# P(X < 7) = sum P(X = k) for k = 0 to 6, where P(X = k) = (e^(-12) * 12^k) / k!

from math import exp, factorial

def poisson_prob(k, lambd):
    return (exp(-lambd) * lambd**k) / factorial(k)

p_fewer_than_7 = sum(poisson_prob(k, 12) for k in range(7))

p_fewer_than_7
```

# 4.

```python
# Binomial

# a. P(X = 0) = (C(5,0) * 0.35^0 * 0.65^5)  
# b. P(X < 2) = P(X = 0) + P(X = 1)  
# c. P(X > 3) = P(X = 4) + P(X = 5)  

from math import comb

def binomial_prob(success, trials, prob):
    return comb(trials, success) * (prob**success) * ((1-prob)**(trials-success))

p_none = binomial_prob(0, 5, 0.35)
p_fewer_than_2 = p_none + binomial_prob(1, 5, 0.35)
p_more_than_3 = binomial_prob(4, 5, 0.35) + binomial_prob(5, 5, 0.35)

p_none, p_fewer_than_2, p_more_than_3
```

```python
# Poisson

# a. P(X = 4) = (e^(-6) * 6^4) / 4!  
# b. P(X ≥ 4) = 1 - ∑ P(X = k) for k = 0 to 3  
# c. P(X = 75) = (e^(-72) * 72^75) / 75!  

from math import exp, factorial

def poisson_prob(k, lambd):
    return (exp(-lambd) * lambd**k) / factorial(k)

p_exactly_4 = poisson_prob(4, 6)
p_at_least_4 = 1 - sum(poisson_prob(k, 6) for k in range(4))
p_exactly_75 = poisson_prob(75, 72)

p_exactly_4, p_at_least_4, p_exactly_75
```

```python
# Binomal

# a. P(3 ≤ X ≤ 6) = P(X = 3) + P(X = 4) + P(X = 5) + P(X = 6)  
# b. P(X < 4) = P(X = 0) + P(X = 1) + P(X = 2) + P(X = 3)  
# c. P(X > 5) = P(X = 6) + P(X = 7) + ... + P(X = 15)

p_3_to_6 = sum(binomial_prob(k, 15, 0.25) for k in range(3, 7))
p_fewer_than_4 = sum(binomial_prob(k, 15, 0.25) for k in range(4))
p_more_than_5 = sum(binomial_prob(k, 15, 0.25) for k in range(6, 16))

p_3_to_6, p_fewer_than_4, p_more_than_5
```

```python
# Negative Binomial Distribution

# a. P(X = 3) = (1 - p)^(3-1) * p  
# b. P(X = 7) = (1 - p)^(7-3) * p  
# c. Mean = r / p, Variance = r * (1 - p) / p²  


p = 0.2
r = 3

# a. Probability that the first strike comes on the third well
p_first_strike_3 = (1 - p)**(3-1) * p

# b. Probability that the third strike comes on the seventh well
p_third_strike_7 = (1 - p)**(7-3) * p

# c. Mean and variance for 3 strikes
mean = r / p
variance = r * (1 - p) / p**2

p_first_strike_3, p_third_strike_7, mean, variance
```
