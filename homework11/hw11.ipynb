# 1. Credit Card Debt for Graduate Students

```python
mu = 3262
sigma = 1100
```

# 1(a) A randomly selected graduate student owes at least $1000.


```python
from scipy.stats import norm

p_at_least_1000 = 1 - norm.cdf(1000, loc=mu, scale=sigma)
print("P(X >= 1000) =", p_at_least_1000)
```

# 1(b) A randomly selected graduate student owes more than $4000.


```python
p_more_than_4000 = 1 - norm.cdf(4000, loc=mu, scale=sigma)
print("P(X > 4000) =", p_more_than_4000)
```

# 1(c) A randomly selected graduate student owes between $3000 and $4000.


```python
p_between_3000_4000 = norm.cdf(4000, loc=mu, scale=sigma) - norm.cdf(3000, loc=mu, scale=sigma)
print("P(3000 <= X <= 4000) =", p_between_3000_4000)
```

# 5. Committee Selection from Foreign Student Club

```python
from math import comb

canadians = 2
japanese = 3
italians = 5
germans = 2

total_people = canadians + japanese + italians + germans
total_committees = comb(total_people, 4)
```

# 5(a) Find the probability that all nationalities are represented in the committee.


```python
ways_all_represented = comb(canadians, 1) * comb(japanese, 1) * comb(italians, 1) * comb(germans, 1)
p_all_represented = ways_all_represented / total_committees
p_all_represented
```

# 5(b) Find the probability that all nationalities except Italian are represented.


```python
ways_non_italians_valid = (
    comb(canadians, 2) * comb(japanese, 1) * comb(germans, 1)
    + comb(canadians, 1) * comb(japanese, 2) * comb(germans, 1)
    + comb(canadians, 1) * comb(japanese, 1) * comb(germans, 2)
)
p_no_italians = ways_non_italians_valid / total_committees
p_no_italians
```

# 6. Light Bulb Lifespan Probability; Find the probability of exactly four bulbs lasting less than 500 hours and two lasting more than 800 hours in a sample of ten.


```python
from math import factorial

p_A = 0.5  
p_B = 0.2  
p_C = 0.3  

n = 10
a, b, c = 4, 2, 4  


coefficient = factorial(n) // (factorial(a) * factorial(b) * factorial(c))
probability = coefficient * (p_A ** a) * (p_B ** b) * (p_C ** c)

print(f"Probability = {probability}")
```
